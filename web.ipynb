```python
import os

# Ruta principal de tu proyecto
ruta_base = r"C:\Users\bhakt\padregregoriotomas"

# Definir las subcarpetas de origen y de salida
dir_originales = os.path.join(ruta_base, "imagenes_originales")
dir_procesadas = os.path.join(ruta_base, "imagenes_procesadas")

# Crear las carpetas si no existen
os.makedirs(dir_originales, exist_ok=True)
os.makedirs(dir_procesadas, exist_ok=True)

print("¡Estructura de carpetas creada con éxito!")
print(f"1. Pasa tus imágenes originales a: {dir_originales}")
print(f"2. Tu carpeta de salida para la alta calidad será: {dir_procesadas}")
```

```python
from PIL import Image, ImageEnhance  # O la librería que utilices para la mejora

# Listar todas las imágenes en imagenes_originales
extensiones = ('.jpg', '.jpeg', '.png', '.JPG', '.PNG')
archivos = [f for f in os.listdir(dir_originales) if f.endswith(extensiones)]

print(f"Procesando {len(archivos)} imágenes...")

for nombre_archivo in archivos:
    path_entrada = os.path.join(dir_originales, nombre_archivo)
    path_salida = os.path.join(dir_procesadas, nombre_archivo)
    
    # 1. Abrir imagen original
    img = Image.open(path_entrada)
    
    # -------------------------------------------------------------
    # 2. AQUÍ APLICAS TU CÓDIGO DE AUMENTO DE RESOLUCIÓN/MEJORA
    # -------------------------------------------------------------
    # Ejemplo de procesamiento:
    img_procesada = img.convert('RGB')  # O tu algoritmo de super-resolución
    
    # 3. Guardar en la carpeta de salida (output)
    img_procesada.save(path_salida, quality=95)

print("¡Proceso completado! Todas las imágenes procesadas se guardaron en 'imagenes_procesadas'.")
```
